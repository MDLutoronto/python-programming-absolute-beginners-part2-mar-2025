# Compound Data Types - Dictionaries

Like a [list](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#list-in-python), a [dictionary](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#dictionary-in-python) can hold many values within a single [variable](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#variable-in-python). We have seen that the items of a [list](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#list-in-python) are stored in a strictly-ordered fashion, starting from item 0. In a [dictionary](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#dictionary-in-python), each [value](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#keyvalue-pair) is stored in relation to a descriptive [key](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#keyvalue-pair) forming a [key/value pair](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#keyvalue-pair). The most useful aspect of a [dictionary](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#dictionary-in-python) is the ability to supply a [key](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#keyvalue-pair) and receive a [value](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#keyvalue-pair) without reference to [indices](https://docs.constellate.org/key-terms/#index-number). Whereas a [list](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#list-in-python) is typed with brackets `[]`, a [dictionary](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#dictionary-in-python) is typed with braces `{}`. The [key](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#keyvalue-pair) and/or [value](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#keyvalue-pair) can be an [integer](https://docs.constellate.org/key-terms/#integer), [float](https://docs.constellate.org/key-terms/#float), or [string](https://docs.constellate.org/key-terms/#string). Key/value pairs are separated by a colon `:` and separated from other key-value pairs by a comma `,`.


`example_dictionary = {key1 : value1, key2 : value2, key3 : value3}`

We could imagine, for example, a [dictionary](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#dictionary-in-python) with Country names as [keys](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#keyvalue-pair) and their capital cities as [values](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#keyvalue-pair).

````python
# An example of a dictionary countries and capital cities
countries ={
 'Canada': 'Ottawa',
 'Australia': 'Canberra',
 'Germany': 'Berlin',
 'Japan': 'Tokyo',
 'Indonesia': 'Jakarta',
 'France': 'Paris',
 'Argentina': 'Buenos Aires',
 'Denmark': 'Copenhagen',
 'Ghana': 'Accra',
 'Papua New Guinea': 'Port Moresby',
 'Guatemala': 'Guatemala City',
 'Syria': 'Damascus',
 'Sweden': 'Stockholm',
 'Senegal': 'Dakar'}
import pprint # We import the pretty print module so that we can use its pprint function which
# prints out dictionaries in a neater fashion than the built-in print() function
pprint.pprint(countries) # Use the pretty print function to print out `countries`
````

````python
# To look up the value of an entry, specify the key in square brackets after the dictionary variable name
countries['Denmark']
````

````python
# To add a new item to a dictionary, specify the new key in square brackets (as if you were looking it up), but then set it equal to the new value
countries['Finland'] = 'Helsinki'
pprint.pprint(countries)
````

## Dictionary Methods

We'll take a look at six useful [methods](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#method-in-python) for working with [dictionaries](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#dictionary-in-python): `update()`, `pop()`, `keys()`, `values()`, `items()`, and `get()`.

|Method Name | Purpose | Form |
|---|---|---|
|update()| add new key/value pairs to a dictionary | dict_name.update(\[(key1, value1), (key2, value2)])|
| &nbsp; | combine two dictionaries |dict_name.update(dict_name2)|
|pop()| removes a key/value pair from a dictionary | dict_name.pop(key_name) |
|keys()| check if a key is in a dictionary (True/False) | key_name in dict_name.keys()|
| &nbsp; | Loop through the keys in a dictionary | for k in dict_name.keys():|
|values()| check if a value is in a dictionary (True/False) | value_name in dict_name.values()|
| &nbsp;| Loop through the values in a dictionary | for v in dict_name.values():|
|items()| Loop through the keys and values in a dictionary | for k, v in dict_name.items():|
|get()| retrieve the value for a specific key | dict_name.get(key_name) |

### The `update()` Method

The `update()` [method](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#method-in-python) is useful for adding many [key/value pairs](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#keyvalue-pair) to a [dictionary](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#dictionary-in-python) at once. The `update()` [method](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#method-in-python) accepts a single [key/value pair](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#keyvalue-pair), multiple pairs, or even other [dictionaries](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#dictionary-in-python).

````python
# Add a single key/value pair to the dictionary countries using the update() method
countries.update([('Poland', 'Warsaw')])

pprint.pprint(countries) # Use the pretty print function to print `countries'
````

````python
# Adding several key/value pairs to the dictionary countries using the update() method
countries.update([('Philippines', 'Manila'), ('Kenya', 'Nairobi')])

pprint.pprint(countries) # Use the pretty print function to print `countries`
````

### The `pop()` Method

The `pop()` [method](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#method-in-python) is useful for removing [key/value pairs](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#keyvalue-pair) from a [dictionary](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#dictionary-in-python). The `pop()` [method](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#method-in-python) accepts a [key](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#keyvalue-pair), and then removes that [key/value pair](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#keyvalue-pair).

````python
countries.pop('Japan')
pprint.pprint(countries) # Use the pretty print function to print `countries`
````

### The `keys()` and `values()` Methods

The `keys()` and `values()` [methods](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#method-in-python) are useful for when checking whether a particular [key](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#keyvalue-pair) or [value](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#keyvalue-pair) exists in a [dictionary](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#dictionary-in-python). We can pair them with `in` or `not in` operators to check whether a key or value is in our [dictionary](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#dictionary-in-python) - just like we did with [lists](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#list-in-python).

````python
# Loop through all keys in our dictionary
countries.keys()
````

````python
# Checking if the key 'Canada' is in the countries dictionary
'Canada' in countries.values()
````

### The `get()` Method

If we are sure a [key](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#keyvalue-pair) exists, we can use the get method to return it's value.

````python
# Return a value for a particular key
countries.get('Australia')
````

If the [key](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#keyvalue-pair) is not found, no value is displayed. (Optionally, we can also specify a default message to return.)

`dict_name.get('key_name', 'key_not_found_message')`

````python
# Using the get() method to retrieve the value for the key 'Norway'
countries.get('Canada', 'That country is not on our list!')
````

## Flow Control Statements

If a program is a set of steps for accomplishing a task, then [flow control statements](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#flow-control-statement-in-python) help the program decide the next action.

[Flow control statements](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#flow-control-statement-in-python) work like a flowchart. Remember our homework example.

![Flowchart to hangout with friends](https://ithaka-labs.s3.amazonaws.com/static-files/images/tdm/tdmdocs/friends_flowchart.png)

Each diamond in our flowchart represents a decision that has to be made about the best step to take next. This is the essence of [flow control statements](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#flow-control-statement-in-python). They help a program decide what the next step should be given the current circumstances.

## Boolean Values

One way to create [flow control statements](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#flow-control-statement-in-python) is with [boolean values](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#boolean-operator).

[Boolean values](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#boolean-operator) are a  data type that only has two possible values: **True** or **False**. In our flowchart example above, we could consider a "Yes" to be "True" and a "No" to be "False."

When we get the data we need to answer each question, we could then store that answer in a variable and then use that to check which action to take next.

## Comparison Operators
Remember that we can use a [comparison operator](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#comparison-operator-in-python) to help make decisions based on criteria and  values stored in our variables.

## Writing a Flow Control Statement

We use all of these building block to evaluate conditions and write [flow control statements](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#flow-control-statement-in-python) in [Python](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#python-programming-language.



### `if` Statements

An `if` statement begins with an [expression](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#expression-in-python) that evaluates to **True** or **False**.

* if **True**, then perform this action
* if **False**, skip over this action

In practice, the form looks like this:

`if this is True:` <br />
&nbsp; &nbsp; &nbsp; &nbsp; `perform this action`



### `else` Statements

An `else` statement *does not require a condition* to evaluate to **True** or **False**. It simply executes when none of the previous conditions are met. So you would always precede an `else` statement with an `if` statement. An `else` statement nevers sits by itself. The form looks like this:

`else:` <br />
&nbsp; &nbsp; &nbsp; &nbsp; `perform this action`




### `elif` Statements

An `elif` statement, short for "else if," allows us to create a list of possible conditions where one (and only one) action will be executed. `elif` statements come after an initial `if` statement and before an `else` statement:

`if condition A is True:` <br />
&nbsp; &nbsp; &nbsp; &nbsp; `perform action A` <br />
`elif condition B is True:` <br />
&nbsp; &nbsp; &nbsp; &nbsp; `perform action B` <br />
`elif condition C is True:` <br />
&nbsp; &nbsp; &nbsp; &nbsp; `perform action C` <br />
`elif condition D is True:` <br />
&nbsp; &nbsp; &nbsp; &nbsp; `perform action D` <br />
`else:` <br />
&nbsp; &nbsp; &nbsp; &nbsp;`perform action E`

When an `elif` condition is met, all other `elif` statements are skipped over.

Remember our example that brought it all together. Our program asked the user if they are having a good day. It responded to both "Yes" and "No" with unique answers, using `if` and `elif` statements. Then an `else` statement was added that responded to any user input that was not "Yes" or "No".

![Flowchart showing three branches](https://ithaka-labs.s3.amazonaws.com/static-files/images/tdm/tdmdocs/good_day_flowchart3.png)

````python
# A program that responds to whether the user is having a good or bad day
having_good_day = input('Are you having a good day? (Yes or No) ') # Use having_good_day to hold the user's input

if having_good_day == 'Yes': # If the user typed in 'Yes'
    print('Glad to hear your day is going well!') # Print: Glad to hear your day is going well!

# An elif statement for when the user typed in 'No'
elif having_good_day == 'No':
    print('Sorry to hear that. I hope it gets better.')

# An else statement that catches if the answer is not 'Yes' or 'No'
else: # Execute this if none of the other branches executes
    print('Sorry, I only understand "Yes" or "No"')
    # Note that we can use double quotations in our string because it begins and ends with single quotes
````

# Boolean Operators (and/or/not)
But what if we have multiple expressions to check? In our example above, what if the user types "yes" with a lowercase "y". Our program wouldn't recognize that as "yes" and would instead print *Sorry, I only understand "Yes" or "No"* because that expression check is case sensitive. What we want is for the program to check if `having_good_day == "Yes"` or `having_good_day == "yes"`. Either way the user writes the word "yes", the program will print out the correct response of *Glad to hear your day is going well*.

To handle this type of scenario, we can use [boolean operators](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#boolean-operator) (**and**, **or**, or **not**) to combine [expressions](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#expression-in-python) and create larger [expressions](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#expression-in-python) that evaluate to a single [boolean value](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#boolean-operator) (**True**/**False**).

Let's take a look at [boolean operators](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#boolean-operator) in more detail. You may have encountered [boolean operators](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#boolean-operator) before when searching for items in a database or on the web. Let's say you were searching for pictures that had cute pigs and kittens cuddling together in them. You would write a search that looks something like `pigs and kittens`. You don't just want a picture of a pig or a picture of a kitten. Both animals have to be in one picture! You can see that [boolean operators](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#boolean-operator) can match how we would say what we want in plain English.

Note, though, that Python only recognizes these [boolean operators](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#boolean-operator) when they are all lower case, as opposed to [boolean values](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#boolean-operator) that have to start with a capital letter.

#### Using the Boolean Operator `and`
The `and` operator determines whether *both* conditions are **True**.

Let’s say that I’m looking at a photo, and I want to evaluate if it meets my criteria of having both a pig **and** a kitten in it. So let’s check both of these conditions and see if it meets my needs. If both conditions are true, then the whole condition should be true and this is a photo that I want. Let’s check.


````python
# If condition one is True AND condition two is True
# What will the evaluation be?
True and True
````

But what if both conditions were not true. There was a cute pig in the photo, but no kitten in sight. Then this isn’t a photo I’m looking for and the whole condition should be false. Let’s check.

````python
# If condition one is True AND condition two is False
# What will the evaluation be?
True and False
````

In order for an ```and``` [expression](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#expression-in-python) to evaluate to **True**, every condition must be **True**. Here is the "Truth Table" for every pair:

|Expression|Evaluation|
|---|---|
|True and True|True|
|True and False|False|
|False and True|False|
|False and False|False|

Since `and` [expressions](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#expression-in-python) require all conditions to be **True**, they can easily result in **False** evaluations. They are narrower and less forgiving.

#### Using the Boolean Operator ```or```
The ```or``` operator determines whether *any* condition is **True**.

Let’s say I decide to be more open in my search for a cute animal photo. Now I’m okay with a photo as long as it has a pig **or** a kitten. So as long as one condition is true, the whole condition should be true. Let’s check.

````python
# What will the evaluation be here?
True or False
````

I’m more open, but there has to be a pig **or** a kitten in the photo. I’m not going to accept photos of other animals. So if both conditions are false, the whole condition should be false. Let’s check.

````python
# What will the evaluation be here?
False or False
````

An ```or``` [expression](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#expression-in-python) evaluates to **True** if *any* condition is **True**. Here is the "Truth Table" for every pair:

|Expression|Evaluation|
|---|---|
|True or True|True|
|True or False|True|
|False or True|True|
|False or False|False|

Since ```or``` [expressions](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#expression-in-python) only require a single condition to be **True**, they can easily result in **True** evaluations. They are more open and inclusive.

### Using the Boolean Operator ```not```
The```not``` operator only operates on a single expression, essentially flipping **True** to **False** or **False** to **True**.

````python
# The not operator flips a True to False
not True
````

### Combining Boolean and Comparison Operators

So in the end, we can combine [boolean operators](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#boolean-operator) and [comparison operators](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#comparison-operator-in-python) to create even more complicated **Truth** tests or conditions to test to determine the next step in a program.

````python
# Evaluating two conditions with integers at once
(3 < 22) and (60 == 34) # What does each condition evaluate to?
# What does the whole statement evaluate to?
````

````python
# Evaluating two conditions with integers at once
(3 == 45) or (3 != 7) or (1 == 1) # What does each condition evaluate to?
# What does the whole statement evaluate to?
````

You can see that using parentheses to group our conditions helps make the code more readable, but even if we remove the parentheses, the code would still execute.

So far, we have evaluated one or two conditions at once, but we could compare even more, if we wanted, by connecting more [expressions](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#expression-in-python) with [boolean operators](https://github.com/MDLutoronto/text-analysis-glossary/blob/main/README.md#boolean-operator). (In practice, though, this is less common since it creates code that can be difficult to read.)

## String Methods

We can also use methods to work with strings, similar to how we used methods on our lists and dictionaries.

Some common methods of manipulating strings can be found in the table below:

|Code|Results|

|'w'|Opens a file for writing|
|'x'|Creates the specified file, returns an error if the file exists|
